```python
# 以下のコードを貼り付けます。
import numpy as np

x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

import matplotlib.pyplot as plt

# y = ax + b の近似直線(一次関数)導出メソッド
# pを np.polyfit で一次式指定のため最後は1
p = np.polyfit(x, y, 1)

# np.poly1d で p をオブジェクト化 = f
f = np.poly1d(p)

# 散布図作成、点は青色、表示する文字列は score
plt.scatter(x, y, color='blue', label='score')

# 近似直線作成、線は赤色、表示する文字列は lsm
plt.plot(x, f(x), color='red', label='lsm')

# 凡例は左上、軸のラベルは english 、縦軸のラベルは math
plt.legend(loc='upper left')
plt.xlabel('english')
plt.ylabel('math')

plt.show()
```
