```python
import pandas as pd

# データを作成（CSV読み込みの代わり）
data = {
    'name': ['Tanaka', 'Suzuki', 'Sato', 'Nishino', 'Oosawa', 
             'Ishibashi', 'Katagiri', 'Mita', 'Fujisawa', 'Oota'],
    'age': [25, 30, 20, 49, 58, 42, 21, 35, 47, 17],
    'pref': ['Tokyo', 'Saitama', 'Aichi', 'Saga', 'Kouchi', 
             'Tokyo', 'Fukuoka', 'Tokyo', 'Ibaraki', 'Okinawa']
}
df = pd.DataFrame(data)

# 問3の正解コード
# iloc[ 行の範囲 , 列の範囲 ]
# ::2 は「最初から最後まで、2つ飛ばし(step=2)で」という意味
df.iloc[::2, ::2]
```

```python
import pandas as pd

# データを作成（CSV読み込みの代わり）
data = {
    'name': ['Tanaka', 'Suzuki', 'Sato', 'Nishino', 'Oosawa', 
             'Ishibashi', 'Katagiri', 'Mita', 'Fujisawa', 'Oota'],
    'age': [25, 30, 20, 49, 58, 42, 21, 35, 47, 17],
    'pref': ['Tokyo', 'Saitama', 'Aichi', 'Saga', 'Kouchi', 
             'Tokyo', 'Fukuoka', 'Tokyo', 'Ibaraki', 'Okinawa']
}
df = pd.DataFrame(data)

# 問5の正解コード
# 文字列の中で条件を書くのがポイントです
df.query("age >= 35 and pref == 'Tokyo'")
```
