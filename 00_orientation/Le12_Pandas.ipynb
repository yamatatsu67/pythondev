```python
import pandas as pd
import matplotlib.pyplot as plt # グラフのサイズ調整などに必要

# 1. データの読み込み
df = pd.read_csv("../sample_data/california_housing_train.csv")

# 2. 散布図行列を描画
# df[...] : 指定した3つの列だけでグラフ
pd.plotting.scatter_matrix(
    df[['population', 'median_income', 'households']], 
    figsize=(12, 12),  # 【重要】グラフのサイズ (縦, 横)。これがないと小さすぎて見えません
    alpha=0.5,         # 点の透明度 (0~1)。点が重なっても濃淡で密度がわかります
    diagonal='hist',   # 対角線（自分自身）の場所には「ヒストグラム」を表示
    range_padding=0.1  # 枠と点の間隔を少し開ける
)

# 3. グラフを表示
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 色分けやデザインが自動でキレイになります
sns.pairplot(df[['population', 'median_income', 'households']], height=3, diag_kind='hist')
plt.show()
```
