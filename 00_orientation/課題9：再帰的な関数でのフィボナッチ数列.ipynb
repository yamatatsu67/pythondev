```python
import prognum

# キーボードから入力を受け付ける (intで整数に変換)
while True:
    input_n = input("何番目の数値を求めますか？（整数を入力）：")
    
    try:
        n = int(input_n)
        
        if n <= 0:
            print("エラー：1以上の整数を入力してください。")
            continue 

        break

    except ValueError:
        print("エラー：数値（整数）を入力してください。")

# キーボードから入力した数値を引数にして作成した関数で計算する
result = prognum.fibonacci(n)

# 計算結果の数値を画面に出力する
print(f"{n} 番目の数値は {result} です。")
```
