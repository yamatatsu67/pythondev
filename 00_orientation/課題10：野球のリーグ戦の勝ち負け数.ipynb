```python
# Baseball_Team という名前のクラスを作る
# name,win,lose,drawのメンバ変数を持たせる
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

# クラスの中に勝率を算出する calc_win_rate() というメソッドを作る
# 勝率は勝ち数/(勝ち数+負け数) にて算出する
    def calc_win_rate(self):
        # 勝率 = 勝ち / (勝ち + 負け)
        return self.win / (self.win + self.lose)
        
# クラスの中に勝敗情報を表示する show_team_result() というメソッドを作る
    def show_team_result(self):
        rate = self.calc_win_rate()

        # 出力は書式設定を活用してきれいにすることを心掛ける        
        # 縦線 | を入れて表形式にする
        # name : 左寄せ(<) 12文字
        # win, lose, draw : 右寄せ(>) 4文字
        # rate : 小数点3桁(.3f)
        return f"| {self.name:<12} | {self.win:>4} | {self.lose:>4} | {self.draw:>4} | {rate:.3f} |"

# クラスの中に問題文のリーグ戦の勝敗表のインスタンスを生成する
teams = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pythons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2)
]

# 出力結果を例に表のヘッダー（見出し）を表示する
# クラスで定義した表示形式に合わせる
print(f"| {'team':<12} | {'win':>4} | {'lose':>4} | {'draw':>4} | {'rate':<5} |")

# 区切り線を表示して見やすくすることを心掛ける
# クラスで定義した表示形式に合わせる
print(f"|{'-'*14}|{'-'*6}|{'-'*6}|{'-'*6}|{'-'*7}|")

# show_team_result() メソッドを呼び出して、勝敗情報を表示する
for team in teams:
    print(team.show_team_result())
```
